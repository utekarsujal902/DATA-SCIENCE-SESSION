```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, np.nan, 30, np.nan],
    'Salary': [50000, 60000, None, 45000]
}

df = pd.DataFrame(data)
print("Before cleaning")
print(df)
```

```python

df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)

print("After cleaning")
print(df)
```

```python
import numpy as np

data = [10, 12, 13, 12, 11, 200, 14, 13, 15, 12]

Q1, Q3 = np.percentile(data, [25, 75])
IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = [x for x in data if x < lower or x > upper]
print(outliers)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 32, 47],
    'Country': ['US', 'India', 'US'],
    'Spend1': [200, 150, 300],
    'Spend2': [100, 200, 400]
})

df['Total_Spend'] = df['Spend1'] + df['Spend2']
df['Age_Group'] = pd.cut(df['Age'], bins=[0,30,50], labels=['Young','Middle'])

encoder = LabelEncoder()
df['Country_encoded'] = encoder.fit_transform(df['Country'])

print(df)
```

```python
import pandas as pd

df = pd.read_csv("/content/studentsdata.csv")

print(df.head())
print(df.info())
print(df.describe())
print("Missing values:\n", df.isnull().sum())
```

```python

df['Math_Score'] = df['Math_Score'].fillna(df['Math_Score'].mean())
df['Writing_Score'] = df['Writing_Score'].fillna(df['Writing_Score'].median())
df['Parental_Education'] = df['Parental_Education'].fillna(df['Parental_Education'].mode()[0])
df['Reading_Score'] = df['Reading_Score'].fillna(method='ffill')
df = df.dropna(subset=['Attendance'])

print("After cleaning:\n", df.isnull().sum())
```

```python

df['Total_Score'] = df[['Math_Score','Reading_Score','Writing_Score']].sum(axis=1)
df['Average_Score'] = df['Total_Score'] / 3

df['Performance_Level'] = pd.cut(
    df['Average_Score'],
    bins=[0,60,75,100],
    labels=['Low','Medium','High']
)

df['Attendance_Level'] = pd.cut(
    df['Attendance'],
    bins=[0,75,90,100],
    labels=['Poor','Average','Good']
)

print(df.head())
```

```python

print("Average performance by gender")
print(df.groupby('Gender')['Average_Score'].mean())

print("\nPerformance distribution")
print(df['Performance_Level'].value_counts())

print("\nCorrelation between attendance & average score")
print(df[['Attendance','Average_Score']].corr())

print("\nParental education impact")
print(df.groupby('Parental_Education')['Average_Score'].mean())
```
